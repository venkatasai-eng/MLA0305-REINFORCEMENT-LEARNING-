```python
import numpy as np
import random

states = [0, 1, 2, 3]
actions = [0, 1]
gamma = 0.9
alpha = 0.1
epsilon = 0.1
episodes = 500

def step(state, action):
    if state == 3:
        return 3, 0, True

    if action == 0:
        next_state = max(0, state - 1)
    else:
        next_state = min(3, state + 1)

    reward = 10 if next_state == 3 else -1
    done = next_state == 3

    return next_state, reward, done

Q_sarsa = np.zeros((len(states), len(actions)))
Q_qlearning = np.zeros((len(states), len(actions)))

def choose_action(Q, state):
    if random.random() < epsilon:
        return random.choice(actions)
    return np.argmax(Q[state])

sarsa_rewards = []
qlearning_rewards = []

for episode in range(episodes):
    state = 0
    action = choose_action(Q_sarsa, state)
    total_reward = 0

    while True:
        next_state, reward, done = step(state, action)
        next_action = choose_action(Q_sarsa, next_state)

        Q_sarsa[state, action] += alpha * (
            reward + gamma * Q_sarsa[next_state, next_action]
            - Q_sarsa[state, action]
        )

        total_reward += reward
        state = next_state
        action = next_action

        if done:
            break

    sarsa_rewards.append(total_reward)

for episode in range(episodes):
    state = 0
    total_reward = 0

    while True:
        action = choose_action(Q_qlearning, state)
        next_state, reward, done = step(state, action)

        Q_qlearning[state, action] += alpha * (
            reward + gamma * np.max(Q_qlearning[next_state])
            - Q_qlearning[state, action]
        )

        total_reward += reward
        state = next_state

        if done:
            break

    qlearning_rewards.append(total_reward)

print("TD(0) Value Function:")

V = np.zeros(len(states))

for episode in range(episodes):
    state = 0

    while True:
        action = choose_action(Q_qlearning, state)
        next_state, reward, done = step(state, action)

        V[state] += alpha * (
            reward + gamma * V[next_state] - V[state]
        )

        state = next_state

        if done:
            break

print(np.round(V, 3))

print("\nSARSA Q-Table:")
print(np.round(Q_sarsa, 3))

print("\nQ-Learning Q-Table:")
print(np.round(Q_qlearning, 3))

print("\nSARSA Optimal Policy:")

for state in states:
    if state == 3:
        print("State", state, ": Goal")
    else:
        action = "Left" if np.argmax(Q_sarsa[state]) == 0 else "Right"
        print("State", state, ":", action)

print("\nQ-Learning Optimal Policy:")

for state in states:
    if state == 3:
        print("State", state, ": Goal")
    else:
        action = "Left" if np.argmax(Q_qlearning[state]) == 0 else "Right"
        print("State", state, ":", action)
```
