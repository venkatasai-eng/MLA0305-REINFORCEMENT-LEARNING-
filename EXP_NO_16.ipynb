```python
import numpy as np

np.random.seed(42)

states = 5
actions = 2
gamma = 0.9
iterations = 100

def step(state, action):
    next_state = min(state + 1, 4) if action == 1 else max(state - 1, 0)
    reward = 10 if next_state == 4 else -1
    return next_state, reward

V = np.zeros(states)

for _ in range(iterations):
    new_V = np.zeros(states)

    for state in range(states - 1):
        values = []

        for action in range(actions):
            next_state, reward = step(state, action)
            value = reward + gamma * V[next_state]
            values.append(value)

        new_V[state] = max(values)

    V = new_V

policy = np.zeros(states, dtype=int)

for state in range(states - 1):
    values = []

    for action in range(actions):
        next_state, reward = step(state, action)
        values.append(
            reward + gamma * V[next_state]
        )

    policy[state] = np.argmax(values)

print("Value Function:")
print(np.round(V, 2))

print("\nOptimal Policy:")

for state in range(states):
    if state == 4:
        print("State", state, ": Goal")
    else:
        print(
            "State", state, ":",
            "Left" if policy[state] == 0 else "Right"
        )

print("\nValue-Equivalence Prediction Completed.")
```
