```python
import numpy as np

# Grid size
grid_size = 4

# Start and Goal
state = (0, 0)
goal = (3, 3)

reward = 0

print("Starting State:", state)

while state != goal:

    print("\nCurrent State:", state)

    action = np.random.choice(["up","down","left","right"])

    print("Action:", action)

    x, y = state

    if action == "up":
        x = max(0, x-1)

    elif action == "down":
        x = min(grid_size-1, x+1)

    elif action == "left":
        y = max(0, y-1)

    elif action == "right":
        y = min(grid_size-1, y+1)

    state = (x, y)

    if state == goal:
        reward = 10
    else:
        reward = -1

    print("Next State:", state)
    print("Reward:", reward)

print("\nGoal Reached!")
```
