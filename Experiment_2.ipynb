```python
import numpy as np
import random

# States: 0, 1, 2, 3
# Grid:
# 0  1
# 2  3 (Goal)

states = [0, 1, 2, 3]
actions = ["up", "down", "left", "right"]

# Q-table
Q = np.zeros((len(states), len(actions)))

# Parameters
alpha = 0.8      # Learning rate
gamma = 0.9      # Discount factor
epsilon = 0.2    # Exploration rate
episodes = 100

# Environment
def step(state, action):
    if state == 3:
        return state, 10, True

    if action == "up":
        next_state = state - 2 if state >= 2 else state
    elif action == "down":
        next_state = state + 2 if state < 2 else state
    elif action == "left":
        next_state = state - 1 if state % 2 != 0 else state
    elif action == "right":
        next_state = state + 1 if state % 2 == 0 else state

    reward = 10 if next_state == 3 else -1
    done = next_state == 3

    return next_state, reward, done


# Q-Learning
for episode in range(episodes):

    state = 0

    for step_count in range(20):

        # Epsilon-greedy policy
        if random.random() < epsilon:
            action_index = random.randint(0, 3)
        else:
            action_index = np.argmax(Q[state])

        action = actions[action_index]

        next_state, reward, done = step(state, action)

        # Q-learning update
        Q[state, action_index] = Q[state, action_index] + alpha * (
            reward + gamma * np.max(Q[next_state]) - Q[state, action_index]
        )

        state = next_state

        if done:
            break


# Display learned Q-table
print("Learned Q-Table:")
print(Q)

# Extract optimal policy
print("\nOptimal Policy:")

for state in states:
    if state == 3:
        print("State", state, ": GOAL")
    else:
        best_action = actions[np.argmax(Q[state])]
        print("State", state, ":", best_action)


# Test the learned agent
print("\nAgent Path:")

state = 0
path = [state]

for _ in range(10):

    if state == 3:
        break

    action_index = np.argmax(Q[state])
    action = actions[action_index]

    state, reward, done = step(state, action)
    path.append(state)

print(" -> ".join(map(str, path)))
```
