```python
import numpy as np

np.random.seed(42)

states = 5
actions = 2
episodes = 100
gamma = 0.9

def environment(state, action):
    next_state = min(state + 1, 4) if action == 1 else max(state - 1, 0)
    reward = 10 if next_state == 4 else -1
    return next_state, reward

transition = np.zeros((states, actions, states))
rewards = np.zeros((states, actions))

for _ in range(episodes):
    state = np.random.randint(states - 1)
    action = np.random.randint(actions)
    next_state, reward = environment(state, action)

    transition[state, action, next_state] += 1
    rewards[state, action] += reward

for s in range(states):
    for a in range(actions):
        total = np.sum(transition[s, a])
        if total > 0:
            transition[s, a] /= total
            rewards[s, a] /= total

V = np.zeros(states)

for _ in range(100):
    new_V = np.zeros(states)

    for s in range(states - 1):
        values = []

        for a in range(actions):
            value = rewards[s, a] + gamma * np.sum(
                transition[s, a] * V
            )
            values.append(value)

        new_V[s] = max(values)

    V = new_V

policy = np.argmax(
    rewards + gamma * np.sum(
        transition * V.reshape(1, 1, -1),
        axis=2
    ),
    axis=1
)

print("Learned Value Function:")
print(np.round(V, 2))

print("\nOptimized Policy:")

for s in range(states):
    if s == 4:
        print("State", s, ": Goal")
    else:
        print(
            "State", s, ":",
            "Left" if policy[s] == 0 else "Right"
        )

print("\nModel-Based Policy Optimization Completed.")
```
