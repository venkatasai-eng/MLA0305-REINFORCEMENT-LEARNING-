```python
import numpy as np

states = [0, 1, 2, 3]
actions = [0, 1]

gamma = 0.9
def step(state, action):
    if state == 3:
        return 3, 0

    if action == 0:
        next_state = max(0, state - 1)
    else:
        next_state = min(3, state + 1)

    reward = 10 if next_state == 3 else -1

    return next_state, reward


V = np.zeros(len(states))

for iteration in range(100):

    new_V = np.zeros(len(states))

    for state in states:

        if state == 3:
            continue

        action_values = []

        for action in actions:
            next_state, reward = step(state, action)
            value = reward + gamma * V[next_state]
            action_values.append(value)

        new_V[state] = max(action_values)

    if np.max(np.abs(new_V - V)) < 0.001:
        V = new_V
        break

    V = new_V

value_policy = []

for state in states:

    if state == 3:
        value_policy.append("Goal")
        continue

    action_values = []

    for action in actions:
        next_state, reward = step(state, action)
        action_values.append(reward + gamma * V[next_state])

    value_policy.append("Left" if np.argmax(action_values) == 0 else "Right")


print("VALUE ITERATION")
print("Value Function:", np.round(V, 3))
print("Optimal Policy:", value_policy)
print("Iterations:", iteration + 1)



policy = np.zeros(len(states), dtype=int)
for iteration in range(100):

    for _ in range(100):

        new_V = np.zeros(len(states))

        for state in states:

            if state == 3:
                continue

            action = policy[state]
            next_state, reward = step(state, action)

            new_V[state] = reward + gamma * V[next_state]

        V = new_V

    policy_stable = True

    for state in states:

        if state == 3:
            continue

        old_action = policy[state]

        action_values = []

        for action in actions:
            next_state, reward = step(state, action)
            action_values.append(reward + gamma * V[next_state])

        policy[state] = np.argmax(action_values)

        if old_action != policy[state]:
            policy_stable = False

    if policy_stable:
        break


policy_names = []

for state in states:
    if state == 3:
        policy_names.append("Goal")
    elif policy[state] == 0:
        policy_names.append("Left")
    else:
        policy_names.append("Right")


print("\nPOLICY ITERATION")
print("Value Function:", np.round(V, 3))
print("Optimal Policy:", policy_names)
print("Iterations:", iteration + 1)
```
