```python
import numpy as np

# States
states = [0, 1, 2, 3]

# Actions
actions = ["left", "right"]

# Parameters
gamma = 0.9
iterations = 10

def step(state, action):
    if state == 3:
        return 3, 10

    if action == "right":
        next_state = min(state + 1, 3)
    else:
        next_state = max(state - 1, 0)

    reward = 10 if next_state == 3 else -1

    return next_state, reward

V = np.zeros(len(states))

print("State-Value Convergence:")

for i in range(iterations):

    new_V = np.zeros(len(states))

    for state in states:
        if state == 3:
            new_V[state] = 10
            continue

        values = []

        for action in actions:
            next_state, reward = step(state, action)
            values.append(reward + gamma * V[next_state])

        new_V[state] = max(values)

    V = new_V

    print("Iteration", i + 1, ":", np.round(V, 3))


Q = np.zeros((len(states), len(actions)))

print("\nAction-Value Convergence:")

for i in range(iterations):

    new_Q = np.zeros_like(Q)

    for state in states:

        if state == 3:
            continue

        for a, action in enumerate(actions):

            next_state, reward = step(state, action)

            new_Q[state, a] = reward + gamma * np.max(Q[next_state])

    Q = new_Q

    print("Iteration", i + 1, ":\n", np.round(Q, 3))


print("\nFinal State-Value Function:")
print(np.round(V, 3))

print("\nFinal Action-Value Function:")
print(np.round(Q, 3))
```
