```python
import numpy as np
import random

# States
states = [0, 1, 2, 3]

# Actions
actions = [0, 1]   # 0 = Left, 1 = Right

gamma = 0.9
episodes = 500

# Environment
def step(state, action):
    if state == 3:
        return 3, 0, True

    if action == 0:       # Left
        next_state = max(0, state - 1)
    else:                 # Right
        next_state = min(3, state + 1)

    reward = 10 if next_state == 3 else -1
    done = next_state == 3

    return next_state, reward, done


# =========================================================
# MONTE CARLO PREDICTION
# =========================================================

V = np.zeros(len(states))
returns = {state: [] for state in states}

# Fixed policy: move right
def policy(state):
    return 1


for episode in range(episodes):

    state = 0
    episode_data = []

    while True:

        action = policy(state)
        next_state, reward, done = step(state, action)

        episode_data.append((state, reward))
        state = next_state

        if done:
            break

    # Calculate returns
    G = 0

    for state, reward in reversed(episode_data):

        G = reward + gamma * G

        if state not in [s for s, r in episode_data[:-1]]:
            returns[state].append(G)
            V[state] = np.mean(returns[state])


print("MONTE CARLO PREDICTION")
print("Estimated State Values:")
print(np.round(V, 3))


# =========================================================
# MONTE CARLO CONTROL
# =========================================================

Q = np.zeros((len(states), len(actions)))
returns_Q = {(s, a): [] for s in states for a in actions}

epsilon = 0.1


def choose_action(state):

    if random.random() < epsilon:
        return random.choice(actions)

    return np.argmax(Q[state])


for episode in range(episodes):

    state = 0
    episode_data = []

    # Generate episode
    while True:

        action = choose_action(state)

        next_state, reward, done = step(state, action)

        episode_data.append((state, action, reward))

        state = next_state

        if done:
            break

    # Calculate returns
    G = 0
    visited = set()

    for state, action, reward in reversed(episode_data):

        G = reward + gamma * G

        if (state, action) not in visited:

            visited.add((state, action))

            returns_Q[(state, action)].append(G)

            Q[state, action] = np.mean(
                returns_Q[(state, action)]
            )


# Extract optimal policy
optimal_policy = []

for state in states:

    if state == 3:
        optimal_policy.append("Goal")
    elif np.argmax(Q[state]) == 0:
        optimal_policy.append("Left")
    else:
        optimal_policy.append("Right")


print("\nMONTE CARLO CONTROL")
print("Action-Value Function:")
print(np.round(Q, 3))

print("\nOptimal Policy:")
print(optimal_policy)
```
