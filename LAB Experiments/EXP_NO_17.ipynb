```python
import numpy as np

np.random.seed(42)

states = 5
gamma = 0.9

def move(state, direction):
    if direction == "left":
        next_state = max(0, state - 1)
    else:
        next_state = min(states - 1, state + 1)

    reward = 10 if next_state == states - 1 else -1
    return next_state, reward

def move_to_goal(state):
    total_reward = 0

    while state < states - 1:
        state, reward = move(state, "right")
        total_reward += reward

    return state, total_reward

def return_home(state):
    total_reward = 0

    while state > 0:
        state, reward = move(state, "left")
        total_reward += reward

    return state, total_reward

def maxq_task(state, task):
    if task == "ReachGoal":
        return move_to_goal(state)

    if task == "ReturnHome":
        return return_home(state)

    return state, 0

q_reach = np.zeros(states)
q_return = np.zeros(states)

for episode in range(100):

    state = 0
    state, reward = maxq_task(state, "ReachGoal")
    q_reach[state] += reward

    state, reward = maxq_task(state, "ReturnHome")
    q_return[state] += reward

print("MAXQ Hierarchical Learning")
print("Reach Goal Values:")
print(np.round(q_reach, 2))

print("\nReturn Home Values:")
print(np.round(q_return, 2))

print("\nHierarchical Task Decomposition:")
print("Root Task -> ReachGoal -> MoveRight")
print("Root Task -> ReturnHome -> MoveLeft")

print("\nMAXQ Training Completed.")
```
