```python
import numpy as np

np.random.seed(42)

agents = 2
actions = 2
episodes = 100

q = np.zeros((agents, actions))

for episode in range(episodes):

    rewards = np.zeros(agents)

    for agent in range(agents):

        action = np.argmax(q[agent])

        if np.random.rand() < 0.1:
            action = np.random.randint(actions)

        if action == 1:
            rewards[agent] = 10
        else:
            rewards[agent] = -1

        q[agent, action] += 0.1 * (
            rewards[agent] - q[agent, action]
        )

print("Learned Q-Values:")
print(np.round(q, 2))

print("\nOptimal Actions:")

for agent in range(agents):
    action = np.argmax(q[agent])

    print(
        "Agent", agent + 1,
        ": Action", action,
        "Reward:", round(q[agent, action], 2)
    )

print("\nMulti-Agent Reinforcement Learning Completed.")
```
