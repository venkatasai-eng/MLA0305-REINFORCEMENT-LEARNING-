```python
import numpy as np

np.random.seed(42)

states = 5
actions = 2
horizon = 5
samples = 100

def step(state, action):
    if action == 1:
        next_state = min(state + 1, states - 1)
    else:
        next_state = max(state - 1, 0)

    reward = 10 if next_state == states - 1 else -1

    return next_state, reward

best_sequence = None
best_reward = -np.inf

for _ in range(samples):

    state = 0
    sequence = []
    total_reward = 0

    for _ in range(horizon):

        action = np.random.randint(actions)
        sequence.append(action)

        state, reward = step(state, action)
        total_reward += reward

        if state == states - 1:
            break

    if total_reward > best_reward:
        best_reward = total_reward
        best_sequence = sequence

print("Best Action Sequence:")

for action in best_sequence:
    print("Right" if action == 1 else "Left")

print("\nBest Reward:", best_reward)
print("\nSampling-Based Planning Completed.")
```
